```python
import os 
from langgraph.graph import StateGraph, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
# intialize the gemini model
llm= ChatGoogleGenerativeAI( model="gemini-2.5-flash",api_key=os.getenv("GOOGLE_API_KEY"))
```

```python
class BlogState(TypedDict):
    title:str
    outline:str
    content:str
    evaluation:str
```

```python
def create_outline(state:BlogState)->BlogState:

    # fetch the title
    title=state['title']

    # callthe llm  generate outline

    prompt=f'Generate a detailed outline for a blog post with the title: "{title}".'

    outline=llm.invoke(prompt).content

    state['outline']=outline
    return state




    # update state
```

```python
def create_blog(state:BlogState)->BlogState:

    # fetch the title and outline
    title=state['title']
    outline=state['outline']

    # call the llm to generate blog content
    prompt=f'write a detailed blog on the title -{title} using the following  outline \n{outline}'

    content=llm.invoke(prompt).content

    state['content']=content
    return state
```

```python
def evaluate_blog(state:BlogState) -> BlogState:

    # fetch the title and content
    title=state['title']
    content=state['content']
    

    # call the llm to evaluate the blog content
    prompt = f"""
    Evaluate the blog titled '{title}'.
    Give:
    1. Score out of 10
    2. Feedback

    Blog:
    {content}
    """
    evaluation=llm.invoke(prompt).content
    state['evaluation']=evaluation
    return state
```

```python
    # crate a graph
graph=StateGraph(BlogState)

# nodes
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)
graph.add_node('evaluate_blog',evaluate_blog)


# add edges

graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog','evaluate_blog')
graph.add_edge('evaluate_blog',END)

workflow=graph.compile()
```

```python
initial_state={'title':'The rise AI in India'}

final_state=workflow.invoke(initial_state)
print(final_state)
```
